```python
# magic methods
class custom_number:
    def __init__(self, value=0):
        self.value = value
    def __str__(self):
        return f'{self.value}'
    def __add__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value + other.value)  
        else:
            return custom_number(self.value + other)
    def __sub__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value - other.value)  
        else:
            return custom_number(self.value - other)
    def __mul__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value * other.value)  
        else:
            return custom_number(self.value * other)
    def __truediv__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value / other.value)  
        else:
            return custom_number(self.value / other) 
    def __lt__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value < other.value)  
        else:
            return custom_number(self.value < other) 
    def __gt__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value > other.value)  
        else:
            return custom_number(self.value > other) 
    def __floordiv__(self, other):
        if isinstance(other, custom_number):
            return custom_number(self.value // other.value)  
        else:
            return custom_number(self.value // other)             
                   
    def method(self):
        print(f'value is {self.value}')

a = custom_number(10)  
b = custom_number(5)
c = a + b
d = a - b
e = a * b
f = a / b
g = a < b
h = a > b
i = a // b
print(c, type(c))
print(d, type(d))             
print(e, type(e))             
print(f, type(f))
print(g, type(g)) 
print(h, type(h))
print(i, type(i))                         
```

```python
# Diamond Problem ------


class A:
    def disp(self):
        print('print from A')
class B(A):
    def disp(self):
        print('print from B')
        super().disp()
class C(A):
    def disp(self):
        print('print from C')
        super().disp()
class D(B,C):
    def disp(self):
        print('print from D')
        super().disp()

       
obj = D()
D.mro()
#obj.disp()   
```

```python
# MRO
class A1:
    def disp(self):
        print('print from A1')
        super().disp()
class A2:
    def disp(self):
        print('print from A2')        
class B(A1):
    def disp(self):
        print('print from B')
        super().disp()
class C(A2):
    def disp(self):
        print('print from C')
        super().disp()
class D(B,C):
    def disp(self):
        print('print from D')
        super().disp()

       
obj = D()
#D.mro()
obj.disp() 
```

```python
#MRO ( Method Resolution Order )
class A:
    def disp(self):
        print('print from A')
class B(A):
    def disp(self):
        print('print from B')
        super().disp()
class C(A):
    def disp(self):
        print('print from C')
        super().disp()  
class D(B):
    def disp(self):
        print('print from D') 
        super().disp() 
class E(C):
    def disp(self):
        print('print from E') 
        super().disp() 
class F(D,E):
    def disp(self):
        print('print from F')
        super().disp()  

obj =F()
obj.disp()        
                                            
```

```python
# super class or base class or parent class
class Animal:     #superclass or parent class
    def __init__(self, name):
        self.name = name
    def sound(self):
        return 'Generic sound'
class Cat(Animal):     #subclass or child class
    def sound(self):
        return f'{self.name} Meow..'

obj = Cat('kitty')
obj.sound()            
```
