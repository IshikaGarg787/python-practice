```python
# reverse string without using string methods.
# Input string
input_string = "Hello, World!"

# Initialize an empty string for the reversed string
reversed_string = ""

# Loop through each character in the input string
for char in input_string:
    reversed_string = char + reversed_string

# Output the reversed string
print("Reversed string:", reversed_string)
```

```python
# program to check whether a number is a power of 2 or not.
num=int(input('enter the number='))
if num>0:
    while num%2==0:
        num=num /2
    if num==1:
        print('0')
    else:
        print('1')  
else:
    print('1')             
```

```python
# highest common factor
a=int(input('enter the number='))
b=int(input('enter the number'))

div_a=[]
for i in range(1,a+1):
    if a%i==0:
        div_a.append(i)

div_b=[]
for i in range(1,b+1):
    if b%i==0:
        div_b.append(i)

common=list(set(div_a) & set(div_b))
print(max(common))                
```

```python
# print sum of natural number/factorial of natural number.
import math
n= int(input('enter the number:'))
sum1=0
for i in range(1,n+1):
    b = (i/math.factorial(i))
    sum1 += b
sum2=round(sum1,6)    
print(sum2)            
```

```python
# print the sum of series 1^2/1^3 + 3^2/3^3 + 5^2/5^3 +... n.
n=int(input('enter the number='))
sums=0
for i in range(1,n+1):
    term=1/(2*i-1)
    sums+=term
round_sum=round(sums,6)    
print(round_sum)    
```

```python
# letter count
word = "programming"
letter_count = {}

for letter in word:
    if letter in letter_count:
        letter_count[letter] += 1
    else:
        letter_count[letter] = 1

print(letter_count)
# Output: {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}
```

```python
# combine dictionary
dict1 = {}
dict2 = {}

n1 = int(input("How many key-value pairs in first dictionary? "))
for i in range(n1):
    key = input("Enter key: ")
    value = input("Enter value: ")
    dict1[key] = value

n2 = int(input("How many key-value pairs in second dictionary? "))
for i in range(n2):
    key = input("Enter key: ")
    value = input("Enter value: ")
    dict2[key] = value

combined = dict1.copy()
combined.update(dict2)

print("Combined dictionary:", combined)
  
```

```python
# duplicate elements in list
s = input("Enter a list like [2,3,4,5,3,2]: ")
s = s.strip()[1:-1] 
l = s.split(',')
r = [] 
m = []
for i in range(0,len(l)):
    if(l[i] not in r):
        r.append(l[i])
    else:
        m.append(l[i])
        

print(*m)
```
