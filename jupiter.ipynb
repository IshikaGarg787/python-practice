```python
# longest string without repitition.
# Input string
s = "abcdefabcdefghbb"

# Initialize variables
max_len = 0   # To store the length of the longest substring
current_substr = ""  # To store the current substring being checked

# Iterate through each character in the string
for char in s:
    # If the character is already in the current substring, restart the current substring
    if char in current_substr:
        # Find the index of the repeating character and slice the substring from the next character
        current_substr = current_substr[current_substr.index(char) + 1:]
    
    # Add the current character to the substring
    current_substr += char
    
    # Update the maximum length found so far
    max_len = max(max_len, len(current_substr))

# Output the result
print(max_len)  # Output: 3 (for "abc")
```

```python
# find out all the sets of the list.

# Input set
nums = [1, 2, 3]

# Initialize the list to store the subsets
subsets = []

# Initialize a list to store the current subset
current_subset = []

# We start from index 0
start = 0

# The list to store all subsets, including the empty set
subsets.append([])

# We will loop over all possible elements and recursively build subsets
while start < len(nums):
    # We will append to the current_subset one element at a time starting from 'start' index
    for i in range(start, len(nums)):
        current_subset.append(nums[i])  # Include the element in the current subset
        subsets.append(current_subset[:])  # Append a copy of the current subset
        
    # After we finish processing, backtrack by removing the last element
    # This allows us to explore other possible subsets
    start += 1  # Move to the next element and repeat the process
    
# Output all subsets
print(subsets)
```
