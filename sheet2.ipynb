```python
# armstrong number or not
n=int(input('enter the number='))
length=len(str(n))
sum=0
temp=n
while temp>0:
    digit=temp%10
    cube=digit**length
    sum=sum+cube
    temp=temp//10
if sum==n:
    print('armstrong number')
else:
    print('not armstrong number')        
        
```

```python
# remove duplicates.
# Read input from the user
user_input = input("Enter a list of elements (integers and/or strings) separated by commas: ")

# Convert the user input into a list
lst = [eval(x.strip()) for x in user_input.split(',')]

# Create an empty set to track seen elements
seen = set()

# Use a list comprehension to keep elements only if they haven't been seen
unique_list = [x for x in lst if not (x in seen or seen.add(x))]

# Convert the list to a tuple
result = tuple(unique_list)

# Print the result
print("Resulting tuple:", result)
 


```

```python
#twisted prime
number=int(input('enter='))
num1=str(number)
num2=num1[::-1]
num3=int(num2)
for i in range(1,number+1):
    count=0
    for j in range(1,i+1):
        if i%j==0:
            count+=1
for a in range(1,num3+1):
    counts=0
    for b in range(1,a+1):
        if a%b==0:
            counts+=1
if count==2 and counts==2:
    print(f'{number} is a Twisted \nprime')  
else:
    print(f'{number} is not a Twisted \nprime')                          

```

```python
#question even odd game
num= input('enter=')
even_sum=0
odd_sum=0
for i, digit in enumerate(num):
    digit=int(digit)
    if (i+1) % 2 == 0:
        even_sum += digit
    else:
        odd_sum += digit
difference = even_sum - odd_sum        
if difference % 2 == 0:
    print('false')
else:
    print('true')            

```

```python
# priya text transformation adventure
s = input('enter=')
result = ""
upper = True

for char in s:
    if char.isalpha():
        if upper:
            result += char.upper()
        else:
            result += char.lower()
        
    else:
        result += char
    upper = not upper    

print(result)


```
